```python
import pandas as pd
import csv
import re

df = pd.read_csv("all_reviews_with_predictions.csv")

# Remove line breaks inside review text
df["content"] = df["content"].astype(str).apply(
    lambda x: re.sub(r"[\r\n]+", " ", x)
    )

    # Export a phpMyAdmin-safe CSV
df.to_csv(
    "all_reviews_with_predictions_phpmyadmin.csv",
    index=False,
    encoding="utf-8",
    quoting=csv.QUOTE_ALL,
    escapechar="\\",
    lineterminator="\n"
)

print("✅ Saved: all_reviews_with_predictions_phpmyadmin.csv")
```
